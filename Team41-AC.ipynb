## 1.

```python
import numpy as np
import matplotlib.pyplot as plt

# Given datasets
D1 = np.array([2.8, -0.4, -0.8, 2.3, -0.3, 3.6, 4.1])
D2 = np.array([-4.5, -3.4, -3.1, -3.0, -2.3])

# Define log-likelyhood function
def log_likelyhood(D, theta):
  return np.sum(-np.log(np.pi) - np.log(1 + (D - theta)**2))

# Generate a range of theta values
theta_values = np.linspace(-5, 5, 200)

# Calculate log-likelyhood for each theta value
log_likelyhood_values_D1 = [log_likelyhood(D1, theta) for theta in theta_values]
log_likelyhood_values_D2 = [log_likelyhood(D2, theta) for theta in theta_values]

#plot the log-likelyhood functions
plt.figure(figsize=(10, 5))
plt.plot(theta_values, log_likelyhood_values_D1, label='Log-likelyhood for D1')
plt.plot(theta_values, log_likelyhood_values_D2, label='Log-likelyhood for D2')
plt.xlabel('θ')
plt.ylabel('Log-likelyhood')
plt.title('Log-likelyhood functions for D1 and D2')
plt.legend()
plt.grid()
plt.show()
```

## 2.

```python
# Find the index of the maximum value in log-likelihood for D1 and D2
max_idx_D1 = np.argmax(log_likelyhood_values_D1)
max_idx_D2 = np.argmax(log_likelyhood_values_D2)

# Use these indices to find the corresponding theta values
theta_hat_1 = theta_values[max_idx_D1]
theta_hat_2 = theta_values[max_idx_D2]

print(f"MLE for θ in Class ω1 (D1): θ̂1 = {theta_hat_1}")
print(f"MLE for θ in Class ω2 (D2): θ̂2 = {theta_hat_2}")

# Prior probabilities
p_w1 = len(D1) / (len(D1) + len(D2))
p_w2 = len(D2) / (len(D1) + len(D2))

print(f"Prior for ω1 (No Stress): {p_w1:.2f}")
print(f"Prior for ω2 (Stress): {p_w2:.2f}")

# Define the g function
def g(x, theta1, theta2, p_w1, p_w2):
    log_px_w1 = np.log(1 / (np.pi * (1 + (x - theta1)**2)))
    log_px_w2 = np.log(1 / (np.pi * (1 + (x - theta2)**2)))
    return log_px_w1 - log_px_w2 + np.log(p_w1) - np.log(p_w2)

# Predict function that calculates g for a dataset
def predict(D, theta1, theta2, p_w1, p_w2):
    return [g(x, theta1, theta2, p_w1, p_w2) for x in D]

# Calculate and print g estimations for D1 and D2
g_D1 = predict(D1, theta_hat_1, theta_hat_2, p_w1, p_w2)
g_D2 = predict(D2, theta_hat_1, theta_hat_2, p_w1, p_w2)

print("\nEstimations of g(x) for D1:")
for i, val in enumerate(g_D1):
    print(f"x = {D1[i]}: g(x) = {val:.4f}")

print("\nEstimations of g(x) for D2:")
for i, val in enumerate(g_D2):
    print(f"x = {D2[i]}: g(x) = {val:.4f}")

# Plot g estimations
plt.figure(figsize=(10, 5))
plt.plot(D1, g_D1, 'bo', label='D1 (No Stress)')
plt.plot(D2, g_D2, 'ro', label='D2 (Stress)')
plt.axhline(0, color='k', linestyle='--')
plt.xlabel('x')
plt.ylabel('g(x)')
plt.title('g(x) values for D1 and D2')
# Annotate the points with their g(x) values
for i, (x, gx) in enumerate(zip(D1, g_D1)):
    plt.annotate(f"{gx:.2f}", (x, gx), textcoords="offset points", xytext=(0, 5), ha='center', color='blue')

for i, (x, gx) in enumerate(zip(D2, g_D2)):
    plt.annotate(f"{gx:.2f}", (x, gx), textcoords="offset points", xytext=(0, 5), ha='center', color='red')
plt.legend()
plt.grid()
plt.show()
```

# Μέρος Β

## 1.

```python
import numpy as np
import matplotlib.pyplot as plt

# Given datasets
D1 = np.array([2.8, -0.4, -0.8, 2.3, -0.3, 3.6, 4.1])
D2 = np.array([-4.5, -3.4, -3.1, -3.0, -2.3])

# Define prior probability density function
def prior(theta):
    return 1 / (10 * np.pi * (1 + (theta / 10) ** 2))

# Define likelihood function
def likelihood(D, theta):
    return np.prod(1 / (np.pi * (1 + (D - theta) ** 2)))

# Define posterior using Bayes' theorem

def posterior(D, theta_values):
    prior_values = prior(theta_values)
    likelihood_values = np.array([likelihood(D, theta) for theta in theta_values])
    posterior_values = likelihood_values * prior_values # element-wise multiplication for each theta value
    normalization = np.trapz(posterior_values, theta_values)  # Normalize with trapezoidal rule
    return posterior_values / normalization #the result is an array which contains the posterior for each theta after we take into account D set of data

# Define marginal likelihood for p(x|D)
def marginal_likelihood(x, D, theta_values):
    posterior_values = posterior(D, theta_values)
    px_given_theta = 1 / (np.pi * (1 + (x - theta_values) ** 2))
    return np.trapz(px_given_theta * posterior_values, theta_values) #integrate over theta for all the posterior probabilities multiplied with probability of x given theta

# Generate theta values
theta_values = np.linspace(-10, 10, 1000)

# Compute posteriors
posterior_D1 = posterior(D1, theta_values)
posterior_D2 = posterior(D2, theta_values)

# Plot posteriors
plt.figure(figsize=(10, 5))
plt.plot(theta_values, posterior_D1, label='Posterior for D1')
plt.plot(theta_values, posterior_D2, label='Posterior for D2')
plt.xlabel('θ')
plt.ylabel('Posterior p(θ|D)')
plt.title('Posterior Distributions for D1 and D2')
plt.legend()
plt.grid()
plt.show()
```

## 2.

```python
# Define h(x) function
def h(x, D1, D2, theta_values, p_w1, p_w2):
    log_px_D1 = np.log(marginal_likelihood(x, D1, theta_values))
    log_px_D2 = np.log(marginal_likelihood(x, D2, theta_values))
    return log_px_D1 - log_px_D2 + np.log(p_w1) - np.log(p_w2)

# Prior probabilities
p_w1 = len(D1) / (len(D1) + len(D2))
p_w2 = len(D2) / (len(D1) + len(D2))

# Compute h(x) for D1 and D2
h_D1 = [h(x, D1, D2, theta_values, p_w1, p_w2) for x in D1]
h_D2 = [h(x, D1, D2, theta_values, p_w1, p_w2) for x in D2]

# Print h(x) values
print("\nh(x) for D1:")
for i, val in enumerate(h_D1):
    print(f"x = {D1[i]}: h(x) = {val:.4f}")

print("\nh(x) for D2:")
for i, val in enumerate(h_D2):
    print(f"x = {D2[i]}: h(x) = {val:.4f}")

# Plot h(x) values
plt.figure(figsize=(10, 5))
plt.plot(D1, h_D1, 'bo', label='D1 (No Stress)')
plt.plot(D2, h_D2, 'ro', label='D2 (Stress)')
plt.axhline(0, color='k', linestyle='--')
plt.xlabel('x')
plt.ylabel('h(x)')
plt.title('h(x) values for D1 and D2')
plt.legend()
plt.grid()
plt.show()
```

# Μέρος Γ

## 1η ενότητα

### 1.

```python
from sklearn import datasets
import numpy as np
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier

# loading the dataset
iris = datasets.load_iris()
X = iris.data
y = iris.target
# print ('X is:', X)

# keeping first two columns and data scaling
X = X[:, :2]
X = MinMaxScaler().fit_transform(X) # default feature_range=(0,1)
# print ('X is now:', X)

# splitting according to instructions
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.5, random_state=42)

# preparing our model. we have conducted gridsearch earlier and we kept the best parameters.
# we assumed that we don't need bootstrap, hence it is set to False
rf = RandomForestClassifier(bootstrap = False, criterion='gini', max_depth=5, min_samples_leaf=3,
                            min_samples_split=10, n_estimators=16)
rf.fit(X_train, y_train)

# evaluating our model
y_pred = rf.predict(X_test)
print('Accuracy:', accuracy_score(y_test, y_pred))

# confusion matrix for better overview of the results
confusion_matrix = confusion_matrix(y_test, y_pred)
cm_display = ConfusionMatrixDisplay(confusion_matrix=confusion_matrix,
                                            display_labels=['Setosa', 'Versicolor', 'Virginica'])
cm_display.plot()
plt.show()
```

### 2.

```python
x_min, x_max = X[:, 0].min() - 0.1, X[:, 0].max() + 0.1
y_min, y_max = X[:, 1].min() - 0.1, X[:, 1].max() + 0.1
xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.01),
                      np.arange(y_min, y_max, 0.01))

Z = rf.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)

plt.contourf(xx, yy, Z, alpha=0.5)
plt.scatter(X[:, 0], X[:, 1], c=y, s=20, edgecolor='k')
plt.xlabel('Normalised Sepal Length')
plt.ylabel('Normalised Sepal Width')
plt.title('Decision Boundaries')
plt.show()
```

## 2η ενότητα

### 1.

```python
# setting dataset A
X_train_A, _, y_train_A, _ = train_test_split(X_train, y_train, test_size=0.5, random_state=42)

num_trees = 100
max_depth = 3
gamma = 0.5
# size of each bootstrap sample
n_samples = int(gamma * len(X_train_A))
trees = []
predictions = []
for _ in range(num_trees):
    # randomly choosing samples with replacement = True because of Bootstrap
    indices = np.random.choice(len(X_train_A), n_samples, replace = True)
    X_train_bootstrap = X_train_A[indices]
    y_train_bootstrap = y_train_A[indices]
    dt = DecisionTreeClassifier(max_depth=max_depth)
    dt.fit (X_train_bootstrap, y_train_bootstrap)
    trees.append(dt)

    # making predictions
    y_pred = dt.predict(X_test)
    predictions.append(y_pred)

# majority voting
predictions = np.array(predictions)
final_predictions = np.apply_along_axis(lambda x: np.bincount(x, minlength=len(np.unique(y))).argmax(),
                                        axis=0, arr=predictions)
# evaluating our model
accuracy = accuracy_score(y_test, final_predictions)
print(f'Accuracy for max depth = {max_depth} and gamma = {gamma}: is  {accuracy:.2f}')
```

### 2.

```python
x_min, x_max = X[:, 0].min() - 0.1, X[:, 0].max() + 0.1
y_min, y_max = X[:, 1].min() - 0.1, X[:, 1].max() + 0.1
xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.01),
                      np.arange(y_min, y_max, 0.01))

# grid poins for every point in meshgrid
mesh_predictions = []
for point in np.c_[xx.ravel(), yy.ravel()]:
    tree_preds = np.array([tree.predict(point.reshape(1, -1)) for tree in trees])
    mesh_predictions.append(np.bincount(tree_preds.flatten()).argmax())

Z = np.array(mesh_predictions).reshape(xx.shape)

plt.contourf(xx, yy, Z, alpha=0.5)
plt.scatter(X[:, 0], X[:, 1], c=y, s=20, edgecolor='k')
plt.xlabel('Normalised Sepal Length')
plt.ylabel('Normalised Sepal Width')
plt.title('Decision Boundaries')
plt.show()
```
